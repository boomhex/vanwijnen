```python
import pdfplumber
import os
from pathlib import Path
```

```python
data = Path(
    '../data/03.02_Offertes_ontvangen'
)
list(data.iterdir())
```

```python
def iter_for_pdfs(path: Path):
    pdfs = []
    for path in path.iterdir():
        if path.is_dir():
            new_pdfs = iter_for_pdfs(path)
            if new_pdfs:
                pdfs.extend(new_pdfs)
    pdfs.extend(path.glob("*.pdf"))
    return pdfs
pdfs = iter_for_pdfs(data)
print(pdfs)
```

```python
def read_pdf(path: Path) -> str:
    pages = []
    with pdfplumber.open(path) as pdf:
        for page in pdf.pages:
            text = page.extract_text()
            if text:
                pages.append(text)
    return "\n\n".join(pages)

n_chars = [len(read_pdf(p)) for p in pdfs]
n_chars
```

```python
import numpy as np
n_chars = np.array(n_chars)
n_chars.mean()
```

```python
print(f"Tokens per document: {n_chars.mean() / 4:.2f}")
```
